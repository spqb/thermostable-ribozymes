```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import adabmDCA as abm
```

```python
fname_data = "../experiment_results.csv"
df = pd.read_csv(fname_data)
df.head()
```

```python
df_region = df[df["region"] == "lowDCA_highSB"]
best_mutants_h, best_mutants_s = [], []
for wt_id in list(range(1, 11)):
    wt_id_float = float(wt_id)
    df_wt = df_region[df_region["wt_id"] == wt_id_float]
    best_mutant_row = df_wt.nlargest(1, "activity_60")
    best_mutants_h.append(best_mutant_row["id"].values[0])
    best_mutants_s.append(best_mutant_row["seq"].values[0])
```

```python
abm.write_fasta(
    "../best_mutants.fasta",
    headers=best_mutants_h,
    sequences=best_mutants_s,
)
```
